```python
import soundfile as sf
aud="205009__the5thdimension__powerful-rain-thunder-and-hailstorm.flac"
data, fs = sf.read("205009__the5thdimension__powerful-rain-thunder-and-hailstorm.flac")

print(fs)
print(data.shape)
```

```python
import numpy as np

data = data.mean(axis=1)

print(data.shape)
```

```python
import matplotlib.pyplot as plt

t = np.arange(len(data)) / fs

plt.figure(figsize=(12,4))
plt.plot(t, data)

plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.title("Waveform")

plt.show()
```

```python
start = 2
duration = 1

segment = data[int(start*fs):int((start+duration)*fs)]
```

```python
plt.figure(figsize=(12,4))

plt.plot(segment)

plt.title("1 Second Segment")

plt.xlabel("Samples")
plt.ylabel("Amplitude")

plt.show()
```

```python
from IPython.display import Audio

Audio(segment, rate=fs)
```

```python
from scipy.signal import butter, filtfilt
```

```python
cutoff = 1000

b, a = butter(4, cutoff/(fs/2), btype='low')
```

```python
filtered = filtfilt(b, a, segment)
```

```python
from IPython.display import Audio

Audio(filtered, rate=fs)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio
```

```python
fs = 44100

t = np.linspace(0, 1, fs)
```

```python
y1 = np.sin(2 * np.pi * 440 * t)
```

```python
y2 = np.cos(2 * np.pi * 880 * t)
```

```python
mix = y1 + y2
```

```python
Audio(mix, rate=fs)
```

```python
plt.figure(figsize=(12,4))

plt.plot(mix[:2000])

plt.title("Compound Signal")

plt.show()
```

```python
def stretch(sound, factor):

    stretched = np.interp(
        np.arange(0, len(sound), factor),
        np.arange(len(sound)),
        sound
    )

    return stretched
```

```python
slow = stretch(segment, 0.5)

Audio(slow, rate=fs)
```

```python
fast = stretch(segment, 2)

Audio(fast, rate=fs)
```

```python
newfast=stretch(data,2)
Audio(newfast, rate=fs)
```
