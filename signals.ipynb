```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio
```

```python
fs = 1000          # samples per second
t = np.linspace(0, 2, fs)

freq = 7          # frequency
y = np.sin(2 * np.pi * freq * t)
plt.plot(y)
plt.show()
```

```python
plt.plot(t, y)
plt.show()
```
