```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
from scipy.fft import fft, ifft, fftfreq
from IPython.display import Audio
```

```python
fs = 44100
t = np.linspace(0,1,fs)
y = signal.square(2*np.pi*440*t)
Audio(y, rate=fs)
```

```python
hs = fft(y)
fspec = fftfreq(len(y), 1/fs)
```

```python
plt.figure(figsize=(12,4))
plt.plot(fspec[:2000], np.abs(hs[:2000]))
plt.title("Original Spectrum")
plt.show()
```

```python
hs[0] = 0 # division by zero not possible
```

```python
for i in range(1, len(hs)):
    hs[i] = hs[i] / abs(fspec[i])
```

```python
plt.figure(figsize=(12,4))
plt.plot(fspec[:2000], np.abs(hs[:2000]))
plt.title("Modified Spectrum")
plt.show()
```

```python
newy = ifft(hs)
Audio(np.real(newy), rate=fs)
```

```python
fs = 44100
t = np.linspace(0,1,fs)
y = signal.sawtooth(2*np.pi*440*t)
```

```python
hs = fft(y)
freqs = fftfreq(len(y), 1/fs)
```

```python
hs[0] = 0
for i in range(1, len(hs)):
    hs[i] = hs[i] / abs(freqs[i])
```

```python
newy = ifft(hs)
```

```python
Audio(np.real(newy), rate=fs)
```
